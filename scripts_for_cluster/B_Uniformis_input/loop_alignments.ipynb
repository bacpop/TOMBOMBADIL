```python
import pandas as pd
```

```python
# Read file
file = "/Users/sanchezh/Desktop/tombombadil/TOMBOMBADIL/pangenome_summary.tsv"

# Create a table
og_table = pd.read_csv(file, sep="\t", header=0)

# Filter: keep only rows where the 3rd column >= 0.05
# (column index 2 because Python counts from 0) 
filt_table = og_table[og_table.iloc[:, 2] >= 0.05]  # New filtered table

# Save the new filtered table
filt_table.to_csv("filtered_pangeonome_summary.tsv", sep="\t", index=False)
print(filt_table)
```

```python
#Here I want to create a new column (1st column) that enumerates the rows. I also want to delete the three last columns of the table and add a new last column that contains the exact location of the files.
filt_table.insert(0, 'Row_number', range(1, len(filt_table) + 1)) 

# Rename the second column to 'Gene_Name'
filt_table.columns.values[1] = 'Gene_Name'  # Index 1 = second column

loc_table = filt_table.iloc[:, :-3] #Delete the last three columns, new location table

# Add a new last column with the exact location of the files
loc_table['File_Location'] = "/Users/sanchezh/Desktop/tombombadil/TOMBOMBADIL" + loc_table.iloc[:, 0].astype(str)  # Adjust path as needed

# Save the modified table to a new file
loc_table.to_csv("modified_pangeonome_summary.tsv", sep="\t", index=False)

print(loc_table)
```
